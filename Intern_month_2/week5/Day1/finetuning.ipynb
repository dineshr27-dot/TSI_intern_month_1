1. What is Fine-Tuning?

A simple definition:

Fine-tuning is taking an already pretrained model and training it further on a smaller, task-specific dataset so that its parameters adapt to the desired task or behavior.

Think:

Huge general training
        ↓
    Base LLM
        ↓
Your specific dataset
        ↓
   Fine-tuning
        ↓
Specialized model

For example, a base LLM may already know:

English
Python
mathematics
general knowledge
programming
reasoning patterns

But you may want it to become particularly good at:

Customer support
Medical text classification
Legal response formatting
SQL generation
Code generation
Sentiment classification
Instruction following

Fine-tuning adapts the model to that specific requirement.


Working of Fine-Tuning
Fine-tuning typically involves the following steps

1. Select a Pre-Trained Model
Choose a model that has already been trained on a large and diverse dataset.
Examples include BERT for NLP tasks, ResNet for image classification and GPT models for text generation.

2. Freeze Initial Layers
The early layers are usually kept unchanged because they have already learned general features.
For example, image models learn edges, shapes and textures, while language models learn basic grammar and word relationships.

3. Fine-Tune Later Layers
The later layers are updated using the new dataset.
These layers learn task-specific patterns and adapt the model to the target application.

4. Use a Small Learning Rate
A lower learning rate is used to make gradual adjustments to the model's weights.
This helps preserve previously learned knowledge while allowing the model to adapt to the new task.

5. Evaluate and Refine
The model is tested on the target task to measure its performance.
Based on the results, additional layers can be fine-tuned or training parameters can be adjusted to improve accuracy.

1) Applications

Used to adapt general-purpose models to specific domains such as healthcare, finance and legal services.

Improves performance on specialized tasks like sentiment analysis, question answering and named entity recognition.

Helps models understand specific languages, dialects or writing styles.

Enables personalization based on user preferences, vocabulary or tone.

Allows effective learning from smaller datasets without training a model from scratch.

Supports deployment of optimized models on mobile devices and IoT systems.

2) Advantages

Works well even when only a small amount of training data is available.

Improves performance on domain-specific tasks such as healthcare, finance and legal applications.

Saves time by adapting an existing model instead of training from scratch.

Provides better accuracy by leveraging knowledge learned from large datasets.

Reduces the risk of overfitting on smaller datasets.
Requires fewer computational resources compared to full model training.

3) Limitations

May still suffer from overfitting if the new dataset is too small or lacks diversity.

Can require significant computational resources, especially for large models.

Choosing which layers to freeze and which to fine-tune can be challenging.

Performance depends on the quality and relevance of the pre-trained model.

May not work well when the new task is very different from the original training task

Prompt Engineering 

Prompt engineering is the process of creating clear and effective prompts that guide AI models to generate accurate responses. It mainly focuses on writing smart prompts for text-based AI tasks, especially NLP, to help the user and the model produce the required output.

Prompt Engineering Importance

More specific formats of input as prompts help in better interpretability of the requirements for a task.

Specific prompts with a detailed explanation of the requirements mean output matches more with the desired one.

Better results for NLP tasks, through prompts also mean a better-trained model for future tasks.

# 1. Fine-Tuning vs Prompt Engineering vs RAG

These are three different ways to make an LLM useful for a specific requirement.

## Quick Difference

| Method | What changes? | Training required? | Main purpose |
|---|---|---|---|
| Prompt Engineering | Instructions | No | Tell the model how to respond |
| RAG | Information/context | No model training | Give the model external information |
| Fine-Tuning | Model parameters | Yes | Teach the model new behavior/patterns |

### Easy way to remember

```text
Prompt Engineering
= Tell the model what to do

RAG
= Give the model information

Fine-Tuning
= Train the model to adapt

FINE-TUNING FUNDAMENTALS
========================

1. PROMPT ENGINEERING
2. RAG
3. FINE-TUNING
4. FULL FINE-TUNING
5. PEFT
6. LoRA
7. QLoRA


1. PROMPT ENGINEERING
=====================

What is it?

Prompt Engineering means writing clear instructions to an LLM
to get the desired answer.

It does NOT change the model parameters.

How it works:

User Prompt
    ↓
LLM
    ↓
Answer

Example:

"Explain Python in simple English with 2 examples."

Why use it?

- To control the answer
- To control the format
- To control the style
- To give instructions
- No training is required

Real-life example:

Like telling an employee:

"Answer customers politely and use simple English."

You are giving instructions, not retraining the employee.


2. RAG
======

RAG = Retrieval-Augmented Generation

What is it?

RAG gives an LLM information from external sources
before generating an answer.

The model itself is normally NOT retrained.

How it works:

Documents
    ↓
Chunking
    ↓
Embeddings
    ↓
Vector Database
    ↓
User Question
    ↓
Search relevant information
    ↓
Relevant Chunks
    ↓
LLM
    ↓
Answer

Example:

Company has:

- Company policy
- Product details
- Pricing
- FAQ

User asks:

"What is the company refund policy?"

RAG searches the company documents and gives the relevant
information to the LLM.

Why use RAG?

- Private company information
- PDFs
- Company documents
- Frequently changing information
- Database information
- External knowledge

Important:

RAG does NOT normally change the LLM's parameters.


3. FINE-TUNING
==============

What is it?

Fine-tuning means training a pretrained model further using
specific training data so that it adapts to a particular task
or behavior.

How it works:

Pretrained Model
    ↓
Training Data
    ↓
Prediction
    ↓
Calculate Loss
    ↓
Calculate Gradients
    ↓
Update Parameters
    ↓
Fine-Tuned Model

Example:

A company has thousands of customer-support conversations.

These examples are used to train the model.

The model learns the desired customer-support behavior.

Why use it?

- Specialized tasks
- Specific behavior
- Consistent style
- Classification
- Domain-specific tasks


4. FULL FINE-TUNING
===================

What is it?

Full Fine-Tuning means updating essentially ALL model parameters
during training.

Example:

Model = 7 Billion parameters

Full Fine-Tuning:

7B parameters
████████████████
      ↓
    UPDATE

Why is it expensive?

Because training needs memory for:

- Model weights
- Gradients
- Optimizer states
- Activations

Therefore, large models require a lot of GPU memory.


5. PEFT
=======

PEFT = Parameter-Efficient Fine-Tuning

What is it?

PEFT means fine-tuning a model while training only a small
number of parameters.

Most of the original model is kept frozen.

Full Fine-Tuning:

Base Model
████████████████
      ↓
    TRAIN


PEFT:

Base Model
████████████████
      ↓
    FROZEN

Small Parameters
██
      ↓
    TRAIN


Why use PEFT?

- Less GPU memory
- Less training cost
- Faster training
- Fewer trainable parameters
- Smaller storage requirements

PEFT is a CATEGORY.

Examples:

PEFT
 |
 |-- LoRA
 |-- Prefix Tuning
 |-- Prompt Tuning
 |-- Adapter methods


6. LoRA
=======

LoRA = Low-Rank Adaptation

LoRA is a PEFT technique.

Main idea:

Freeze the original model weights and train small matrices
called A and B.

Normal fine-tuning:

W_new = W + ΔW

LoRA:

W_new ≈ W + BA

W = Original model weights
A = Small trainable matrix
B = Small trainable matrix


What is A?

A is the DOWN-PROJECTION matrix.

It reduces the large dimension to a small dimension.

Example:

4096
  ↓
  A
  ↓
8

A: 4096 → 8


What is B?

B is the UP-PROJECTION matrix.

It increases the small dimension back to the original size.

Example:

8
↓
B
↓
4096

B: 8 → 4096


Together:

Input
  ↓
  A
  ↓
Small representation
  ↓
  B
  ↓
LoRA update


Why does LoRA reduce parameters?

Suppose:

W = 4096 × 4096

Full update:

4096 × 4096
= 16,777,216 parameters


LoRA with rank r = 8:

A = 8 × 4096
  = 32,768

B = 4096 × 8
  = 32,768

Total:

32,768 + 32,768
= 65,536 parameters

So:

Full update = 16,777,216
LoRA        = 65,536

LoRA trains far fewer parameters.


7. RANK IN LoRA
===============

Rank is represented by "r".

Example:

r = 4
r = 8
r = 16
r = 32

Smaller rank:

- Fewer parameters
- Less memory
- Smaller adaptation capacity

Larger rank:

- More parameters
- More memory
- More adaptation capacity

So rank controls the size of the LoRA adapter.


8. LoRA TRAINING
================

Base model:

W
↓
FROZEN

LoRA:

A
↓
TRAINABLE

B
↓
TRAINABLE


During training:

Training Data
    ↓
Model
    ↓
Prediction
    ↓
Loss
    ↓
Gradients
    ↓
Update A and B

The original model weights W remain frozen.


9. QUANTIZATION
===============

What is Quantization?

Quantization means representing model weights using fewer bits.

Examples:

FP32 = 32 bits
FP16 = 16 bits
INT8 = 8 bits
INT4 = 4 bits

Fewer bits
    ↓
Less memory


Why use it?

Large LLMs have billions of parameters.

Using fewer bits reduces the memory needed to store the
model weights.

Example:

FP16
████████████████

INT8
████████

INT4
████


10. QLoRA
=========

QLoRA = Quantized LoRA

QLoRA combines:

Quantization + LoRA


How it works:

Pretrained LLM
    ↓
Quantization
    ↓
Low-bit Base Model
    ↓
Freeze Base Model
    ↓
Add LoRA A + B
    ↓
Train A + B
    ↓
Fine-Tuned Model


In simple words:

Base Model = Quantized + Frozen

LoRA A + B = Trainable


Why use QLoRA?

LoRA reduces the number of trainable parameters.

But the base model still needs to be loaded into GPU memory.

QLoRA also quantizes the base model.

Therefore:

LoRA
↓
Fewer trainable parameters

Quantization
↓
Less memory for base model

QLoRA
↓
Both together


QLoRA is useful when GPU memory is limited.

For example:

Google Colab
Limited GPU memory
        ↓
Quantized model
        +
LoRA
        ↓
More memory-efficient fine-tuning


11. LoRA vs QLoRA
=================

LoRA:

Base Model = Frozen
LoRA A + B = Trainable
Quantization = Not required


QLoRA:

Base Model = Quantized + Frozen
LoRA A + B = Trainable
Quantization = Used


12. PEFT vs LoRA vs QLoRA
=========================

PEFT = Family of efficient fine-tuning methods

LoRA = One PEFT method

QLoRA = Quantization + LoRA


Relationship:

PEFT
 |
 |-- LoRA
 |     |
 |     |-- QLoRA
 |
 |-- Prefix Tuning
 |
 |-- Prompt Tuning


13. MAIN DIFFERENCE
===================

Prompt Engineering
= Change the instructions

RAG
= Give the model external information

Fine-Tuning
= Train the model to adapt

Full Fine-Tuning
= Update essentially all parameters

PEFT
= Update only a small number of parameters

LoRA
= PEFT using A and B matrices

QLoRA
= Quantization + LoRA


14. EASY REAL-LIFE EXAMPLE
==========================

Imagine an employee.

PROMPT ENGINEERING:

Tell the employee:

"Answer customers politely."

You are giving instructions.


RAG:

Give the employee a company handbook.

When they don't know something, they check the handbook.

You are giving information.


FINE-TUNING:

Give the employee thousands of customer-support examples
and train them for customer support.

You are changing their learned behavior.


FULL FINE-TUNING:

Retrain almost everything.


PEFT:

Keep most of their existing knowledge and train only a
small part for the new task.


LoRA:

Add a small adapter that learns the new task.


QLoRA:

Keep the main knowledge in a compressed/quantized form
and train the small LoRA adapter.


15. FINAL MEMORY TRICK
======================

Prompt Engineering
→ Tell

RAG
→ Give information

Fine-Tuning
→ Teach

Full Fine-Tuning
→ Teach everything

PEFT
→ Teach a small part

LoRA
→ Small A + B adapters

QLoRA
→ Quantization + LoRA


MOST IMPORTANT:

PEFT = CATEGORY
LoRA = PEFT METHOD
QLoRA = LoRA + QUANTIZATION
