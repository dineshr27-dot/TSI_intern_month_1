# Prompt Engineering — Versioning, Templates & A/B Testing

## 1. What is a Prompt?

A **prompt** is an instruction given to an LLM to tell it what to do.

Example:

```text
Answer the user's question using the provided context.
```

The LLM receives:

```text
Prompt → LLM → Response
```

Example:

```text
Prompt:
What is RAG?

LLM:
RAG stands for Retrieval-Augmented Generation...
```

---

# 2. What is a System Prompt?

A **system prompt** defines the role, behavior, and rules of the LLM.

Example:

```text
You are a helpful AI assistant.

Rules:
- Answer using the provided context.
- Be accurate.
- Be concise.
- If information is missing, say you don't know.
```

The system prompt controls **how the model should behave**.

### Simple difference

```text
System Prompt
    ↓
How should the AI behave?

User Prompt
    ↓
What does the user want?
```

Example:

```text
SYSTEM:
You are a helpful Python tutor.

USER:
Explain what a list is in Python.
```

---

# 3. Why Do Prompts Need Versioning?

Prompt versioning means keeping different versions of a prompt separately.

Just like code:

```text
Code
V1 → V2 → V3
```

we can have:

```text
Prompt
V1 → V2 → V3
```

Example:

### Prompt V1

```text
Answer the user's question using the provided context.
```

### Prompt V2

```text
Answer the user's question using the provided context.

Be accurate, concise, friendly, and educational.
If the answer is not available in the context, say that the information is not available.
```

The prompt changed from V1 to V2.

We need to keep both versions so we know **what changed**.

---

# 4. Why is Prompt Versioning Important?

Without versioning:

```text
prompt.txt
```

Someone changes the prompt.

Later the application behaves differently.

You don't know:

```text
What changed?
Who changed it?
When was it changed?
Did the change improve the result?
```

With versioning:

```text
prompts/
│
├── prompt_v1.txt
├── prompt_v2.txt
└── prompt_v3.txt
```

Now we can compare them.

---

# 5. What is Prompt Drift?

**Prompt drift** means a prompt gradually changes over time, sometimes without proper tracking or testing.

Example:

Initially:

```text
You are a helpful assistant.
Answer concisely.
```

Someone adds:

```text
Give detailed explanations.
```

Another person adds:

```text
Give examples.
```

Another adds:

```text
Always explain your reasoning.
```

Eventually:

```text
You are a helpful assistant.

Answer concisely.
Give detailed explanations.
Give examples.
Be extremely detailed.
Explain everything.
```

Now the prompt has become complicated.

This gradual uncontrolled change is called **prompt drift**.

---

# 6. Why Prompt Drift Is a Problem?

Prompt drift can change:

* Response length
* Response style
* Accuracy
* Relevance
* Hallucination behavior
* Token usage
* Latency
* User experience

Important:

**These changes are possibilities, not guaranteed results.**

We should test them rather than assume them.

---

# 7. Prompt Versioning vs A/B Testing

These two concepts are different.

### Prompt Versioning

Answers:

> **What changed?**

Example:

```text
V1 → Original prompt
V2 → Added accuracy instructions
```

### A/B Testing

Answers:

> **Did the change improve the result?**

Example:

```text
V1 → Average score = 3.8
V2 → Average score = 4.4

V2 performed better.
```

So:

```text
Versioning
    ↓
Track changes

A/B Testing
    ↓
Measure changes
```

# 8. What Are Variables in Prompts?

A variable is a placeholder whose value changes.

Example:

```text
Answer this question:

{question}
```

Here:

```text
{question}
```

is a variable.

If:

```python
question = "What is RAG?"
```

the final prompt becomes:

```text
Answer this question:

What is RAG?
```

Another question:

```python
question = "What is an embedding?"
```

becomes:

```text
Answer this question:

What is an embedding?
```

The same prompt can therefore be reused.

---

# 9. Why Use Variables?

Without variables:

```text
Answer:
What is RAG?
```

This prompt only works for one question.

With variables:

```text
Answer:
{question}
```

we can use thousands of questions.

```text
Question 1 → {question}
Question 2 → {question}
Question 3 → {question}
...
```

This makes prompts **reusable and maintainable**.

---

# 10. Prompt Template

A prompt template is a reusable prompt containing variables.

Example:

```text
You are a helpful assistant.

Context:
{context}

Question:
{question}

Answer:
```

There are two variables:

```text
{context}
{question}
```

At runtime, our application provides their values.

---

# 11. What is Jinja2?

**Jinja2 is a template engine for Python.**

It allows us to create dynamic templates using variables and simple logic.

Example:

```jinja2
You are a {{ role }}.

Question:
{{ question }}
```

If:

```python
role = "Python tutor"
question = "What is a list?"
```

Jinja2 generates:

```text
You are a Python tutor.

Question:
What is a list?
```

---

# 12. Jinja2 Variable Syntax

Jinja2 commonly uses:

```text
{{ variable }}
```

Example:

```jinja2
Hello {{ name }}
```

If:

```python
name = "Dinesh"
```

result:

```text
Hello Dinesh
```

# 15. What is LangChain PromptTemplate?

LangChain provides a class called:

```python
PromptTemplate
```

It is used to create reusable prompts with variables.

Example:

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["question"],
    template="Answer this question: {question}"
)
```

Then:

```python
result = prompt.format(
    question="What is RAG?"
)
```

Result:

```text
Answer this question: What is RAG?
```

---

# 16. Why Use LangChain PromptTemplate?

It helps us:

```text
Create prompt
      ↓
Define variables
      ↓
Provide values
      ↓
Generate final prompt
```

Instead of manually concatenating strings.

# 17. What is ChatPromptTemplate?

`ChatPromptTemplate` is designed for **chat-based LLM applications**.

It allows us to define different message roles.

Example:

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful assistant."
    ),
    (
        "human",
        "{question}"
    )
])
```

Now we have:

```text
SYSTEM
You are a helpful assistant.

HUMAN
What is RAG?
```

---

# 18. System + Human Messages

A typical chat application looks like:

```text
System Message
       ↓
Defines AI behavior
       ↓
Human Message
       ↓
User's question
       ↓
LLM
       ↓
Response
```

Example:

```text
SYSTEM:
You are a Python teacher.

HUMAN:
Explain functions in Python.
```

# 19. PromptTemplate vs ChatPromptTemplate

| PromptTemplate           | ChatPromptTemplate         |
| ------------------------ | -------------------------- |
| Creates text prompts     | Creates chat messages      |
| Simple prompt structure  | System/Human/AI messages   |
| Good for basic templates | Good for chat applications |
| Uses `{variables}`       | Uses `{variables}`         |
| Text-oriented            | Conversation-oriented      |

Simple rule:

```text
Simple text → PromptTemplate

Chat application → ChatPromptTemplate
```

# 20. Why Use ChatPromptTemplate in Production?

In a real application, prompts can become complex.

For example:

```text
System instructions
+
User question
+
Context
+
Conversation history
+
Additional instructions
```

ChatPromptTemplate helps organize these messages.

Example:

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful assistant. "
        "Use only the provided context."
    ),
    (
        "human",
        "Context:\n{context}\n\nQuestion:\n{question}"
    )
])
```

Then:

```python
messages = prompt.invoke({
    "context": context,
    "question": question
})
```

This is cleaner than manually building a huge string.

# 22. What is A/B Testing?

A/B testing means comparing **two versions** of something.

In our case:

```text
Prompt A = V1
Prompt B = V2
```

We give both prompts the **same test questions**.

---

# 23. Why Use the Same Query Set?

This is very important.

Suppose:

```text
V1 → easy questions

V2 → difficult questions
```

The comparison is unfair.

Instead:

```text
Same 10 questions
        ↓
   ┌────┴────┐
   ↓         ↓
  V1        V2
   ↓         ↓
Answers   Answers
   ↓         ↓
   └────┬────┘
        ↓
     Compare
```

Now the experiment is fairer.

---

# 24. Example A/B Test

Suppose we have:

```text
Q1: What is RAG?
Q2: What is an embedding?
Q3: What is FAISS?
...
Q10: What is hallucination?
```

Run V1:

```text
Q1 → V1 → Answer
Q2 → V1 → Answer
...
Q10 → V1 → Answer
```

Run V2:

```text
Q1 → V2 → Answer
Q2 → V2 → Answer
...
Q10 → V2 → Answer
```

Then compare.

---

# 25. What Should We Compare?

We need evaluation criteria.

### Accuracy

Is the answer correct?

```text
5 = Excellent
1 = Very poor
```

### Relevance

Does the answer actually answer the question?

### Groundedness

For RAG:

> Is the answer supported by the provided context?

### Conciseness

Is the answer unnecessarily long?

### Overall Quality

Overall usefulness of the answer.

# 27. Important A/B Testing Rule

When comparing prompts, keep other things the same.

Use:

```text
Same model
Same questions
Same context
Same temperature
Same generation settings
Same evaluation criteria
```

Only change:

```text
Prompt V1
       vs
Prompt V2
```

This makes the experiment more controlled.
