# LLM OBSERVABILITY & TRACING

## 1. What is Observability?

**Observability** means being able to see and understand what is happening inside an application while it is running.

In an LLM application, many steps may happen internally:

```text
User Query
    ↓
Agent
    ↓
Tool
    ↓
Retriever
    ↓
Vector Database
    ↓
Retrieved Chunks
    ↓
LLM
    ↓
Final Answer
```

Observability helps us see these internal steps.

### Simple definition:

```text
Observability = Visibility into the internal behavior of an application
```

---

# 2. Why is Observability Important?

In a simple application, debugging is easy.

Example:

```python
x = 10
y = 20

result = x + y
print(result)
```

We can easily see what happened.

But an AI application can contain:

```text
Agent
 ↓
Multiple Tools
 ↓
Retriever
 ↓
Vector Database
 ↓
LLM
 ↓
Final Answer
```

If the final answer is wrong, we need to know:

```text
Was the query received correctly?
        ↓
Was the correct tool selected?
        ↓
Did retrieval find the correct chunks?
        ↓
Was the correct context sent to the LLM?
        ↓
Did the LLM generate the correct answer?
        ↓
Which step was slow?
```

Observability helps answer these questions.

---

# 3. Why Observability is Important in RAG

RAG = Retrieval-Augmented Generation.

Basic RAG flow:

```text
User Query
    ↓
Embedding
    ↓
Vector Search
    ↓
Retrieve Chunks
    ↓
Add Chunks to Prompt
    ↓
LLM
    ↓
Answer
```

Suppose the user asks:

```text
"What is the punishment for theft?"
```

The retriever should find:

```text
Chunk 1 → Theft punishment
Chunk 2 → Theft definition
Chunk 3 → Related legal provision
```

But suppose it retrieves:

```text
Chunk 1 → Marriage
Chunk 2 → Divorce
Chunk 3 → Tax
```

This is a **retrieval failure**.

Then:

```text
Bad Retrieval
      ↓
Bad Context
      ↓
LLM
      ↓
Bad Answer
```

Without observability, we may incorrectly blame the LLM.

With observability, we can inspect the retrieved chunks and identify the actual problem.

---

# 4. Why Observability is Important in Agentic Systems

An agent can make decisions and use different tools.

Example:

```text
User
 ↓
Agent
 ↓
Decides what to do
 ↓
Tool
 ↓
Tool Result
 ↓
Agent
 ↓
LLM
 ↓
Final Answer
```

An agent may have:

```text
Tools:
1. Search Tool
2. Calculator
3. Database Search
4. Document Search
```

Example:

```text
User:
"What is 20% of ₹50,000?"

        ↓

Agent

        ↓

Chooses Calculator

        ↓

Calculator

        ↓

₹10,000

        ↓

LLM

        ↓

Final Answer
```

Observability allows us to see:

```text
Which tool was selected?
What input was sent to the tool?
What output did the tool return?
How long did the tool take?
Did the tool fail?
```

---

# 5. What is Tracing?

**Tracing** means recording and following the complete journey of one request through the application.

Example:

```text
User Query
    ↓
Agent
    ↓
Retriever
    ↓
Vector Search
    ↓
Retrieved Chunks
    ↓
LLM
    ↓
Final Answer
```

This complete journey is called a **Trace**.

### Definition:

```text
Trace = Complete journey of one request
```

---

# 6. What is a Span?

A **Span** represents one operation inside a trace.

Example:

```text
Trace
│
├── Agent Span
├── Retriever Span
├── Vector Search Span
├── Tool Span
└── LLM Span
```

Therefore:

```text
Trace = Complete request

Span = One operation inside the request
```

Example:

```text
Trace
   ↓
Agent Span
   ↓
Tool Span
   ↓
LLM Span
```

---

# 7. Trace vs Log

## Log

A log is an individual event recorded by the application.

Example:

```text
Retriever started
Retriever finished
Retrieved 5 chunks
LLM started
LLM finished
```

## Trace

A trace connects all the operations belonging to one request.

```text
User Query
    ↓
Retriever
    ↓
Retrieved Chunks
    ↓
LLM
    ↓
Final Answer
```

Remember:

```text
Log   = Individual event

Trace = Complete request journey
```

---

# 8. What is Latency?

**Latency** means the amount of time taken by an operation.

Example:

```text
User Query
    ↓
Answer returned after 3 seconds
```

Latency:

```text
3 seconds
```

We can also measure each component:

```text
Embedding       → 0.1 sec
Retrieval       → 0.4 sec
LLM             → 2.5 sec
---------------------------
Total           → 3.0 sec
```

Latency helps us identify slow parts of the application.

# 9. What is LLM Observability?

**LLM Observability** means monitoring and understanding the internal behavior of an LLM-based application.

We can observe:

```text
Input Query
Retrieved Documents
Agent Decisions
Tool Calls
LLM Calls
LLM Responses
Latency
Errors
Token Usage
```

Simple definition:

```text
LLM Observability
=
Understanding what happens inside an LLM application
```

# 10. What is LangSmith?

**LangSmith** is a platform from LangChain used for developing, debugging, evaluating, and observing LLM applications.

Think of LangSmith as a **dashboard for inspecting LLM application execution**.

Without observability:

```text
Application
    ↓
LLM
    ↓
Answer
```

With LangSmith tracing:

```text
Application
    ↓
LangSmith
    ↓
Trace
 ├── Input
 ├── Agent
 ├── Tool
 ├── Retriever
 ├── Retrieved Chunks
 ├── LLM
 ├── Response
 └── Latency
```

LangSmith does not replace the RAG or agent.

It **observes and records their execution**.

# 11. What Does "Trace Every Agent Run" Mean?

An **agent run** means one execution of an agent for a particular request.

Example:

```text
User:
"What is bail?"
```

Agent execution:

```text
Agent Run

Input:
"What is bail?"

Decision:
Search legal documents

Tool:
Document Search

Result:
5 chunks

Final Answer:
"Bail is..."
```

The entire execution can be viewed as a trace.

---

# 12. What Does "Trace Every Tool Call" Mean?

Suppose the agent has:

```text
Tools:

search_documents()
calculator()
web_search()
```

User asks:

```text
"What is 25% of ₹40,000?"
```

Agent chooses:

```text
calculator()
```

Trace records:

```text
Tool:
calculator

Input:
40000 * 0.25

Output:
10000

Latency:
0.02 seconds
```

This helps us understand the agent's tool usage.

# 13. What Does "Trace Every LLM Call" Mean?

An application can call the LLM multiple times.

Example:

```text
User
 ↓
Agent
 ↓
LLM Call #1
 ↓
Tool
 ↓
LLM Call #2
 ↓
Final Answer
```

Tracing lets us inspect each LLM call.

Example:

```text
LLM Call #1
Input: ...
Output: ...
Latency: 1.2 sec

LLM Call #2
Input: ...
Output: ...
Latency: 2.0 sec
```

This helps us understand:

```text
Number of LLM calls
Prompt
Response
Latency
Token usage
Errors
```

# 14. What Should We Log?

Important information to record:

```text
1. Input Query
2. Retrieved Chunks
3. Tool Used
4. LLM Response
5. Latency
```

---

## 14.1 Input Query

The original question asked by the user.

Example:

```text
Input Query:
"What is bail?"
```

Why log it?

To know exactly what request started the execution.

---

## 14.2 Retrieved Chunks

Chunks returned by the retriever.

Example:

```text
Query:
"What is bail?"

Retrieved Chunks:

Chunk 1 → Bail definition
Chunk 2 → Bail conditions
Chunk 3 → Court granting bail
```

Why log them?

To check whether the retriever found relevant information.

---

## 14.3 Tool Used

Record which tool the agent selected.

Example:

```text
Agent
 ↓
Selected Tool
 ↓
document_search
```

Why?

To understand agent behavior and detect incorrect tool selection.

---

## 14.4 LLM Response

Record the response generated by the LLM.

Example:

```text
LLM Response:

"Bail refers to the temporary release of an accused..."
```

Why?

To inspect what the model actually generated.

---

## 14.5 Latency

Record how long each operation takes.

Example:

```text
Agent       → 0.2 sec
Retriever   → 0.5 sec
LLM         → 2.0 sec
Total       → 2.7 sec
```

Why?

To identify slow operations.

---

# 15. How to Identify Slow Steps

Suppose the trace shows:

```text
Agent          → 0.2 sec
Embedding      → 0.1 sec
Retrieval      → 5.0 sec
LLM            → 1.5 sec
```

Total:

```text
6.8 seconds
```

We can immediately identify:

```text
Retrieval = Slowest Step
```

Then investigate:

```text
Why is retrieval taking 5 seconds?
```

Possible causes:

```text
Missing vector index
Slow database
Large dataset
Inefficient SQL query
Network latency
```

---

# 16. How to Identify Retrieval Failures

User:

```text
"What is the punishment for theft?"
```

Trace:

```text
Query
 ↓
Retriever
 ↓
Retrieved Chunks

Chunk 1 → Marriage
Chunk 2 → Divorce
Chunk 3 → Tax
Chunk 4 → Employment
Chunk 5 → Property
```

These chunks are not relevant.

Therefore:

```text
❌ Retrieval Failure
```

Then:

```text
Retrieval Failure
      ↓
Wrong Context
      ↓
LLM
      ↓
Wrong Answer
```

The trace helps us find the actual source of the problem.

# 17. RAG Debugging Using Traces

When the answer is wrong, check in this order:

```text
Step 1:
Check Input Query

        ↓

Step 2:
Check Retrieved Chunks

        ↓

Step 3:
Check Prompt / Context

        ↓

Step 4:
Check LLM Response

        ↓

Step 5:
Check Latency

        ↓

Find the Problem
```

Example:

```text
Query
  ↓
Correct ✅
  ↓
Retrieved Chunks
  ↓
Wrong ❌
  ↓
Retrieval Problem
```

Another example:

```text
Query
  ↓
Correct ✅
  ↓
Chunks
  ↓
Correct ✅
  ↓
LLM Response
  ↓
Wrong ❌
  ↓
Investigate prompt/model
```

# 20. Monitoring vs Observability

These are different.

### Monitoring

Answers:

```text
"Is my system healthy?"
```

Example:

```text
Error Rate = 2%
Average Latency = 2 sec
Requests = 10,000
```

### Observability

Answers:

```text
"Why did this particular request fail?"
```

Example:

```text
Request
 ↓
Retriever
 ↓
No relevant chunks
 ↓
Bad context
 ↓
Bad answer
```

Remember:

```text
Monitoring:
What is happening?

Observability:
Why is it happening?
```

# 21. Logs vs Metrics vs Traces

| Concept    | Meaning                             |
| ---------- | ----------------------------------- |
| **Log**    | Individual event                    |
| **Metric** | Numerical measurement               |
| **Trace**  | Complete request journey            |
| **Span**   | Individual operation inside a trace |

Example:

```text
LOG:
"Retriever started"

METRIC:
"Retrieval latency = 500 ms"

TRACE:
Query → Agent → Retriever → LLM → Answer

SPAN:
"Retriever operation"
```
