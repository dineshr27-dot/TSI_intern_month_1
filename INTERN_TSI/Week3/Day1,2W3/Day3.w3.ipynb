```python
# ============================================================
# WEEK 3 - DAY 1: SENTENCE EMBEDDINGS & SIMILARITY
# ============================================================

# Install packages
# Run this once if packages are not installed:
# !pip install sentence-transformers scikit-learn openai voyageai


# ============================================================
# 1. IMPORTS
# ============================================================

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import os


# ============================================================
# 2. LOAD SENTENCE EMBEDDING MODEL
# ============================================================

model = SentenceTransformer("all-MiniLM-L6-v2")


# ============================================================
# 3. SENTENCES
# ============================================================

sentences = [
    "How do I reset my password?",
    "I forgot my password. How can I change it?",
    "What is machine learning?",
    "Python is a programming language."
]

print("Number of sentences:", len(sentences))


# ============================================================
# 4. CREATE EMBEDDINGS
# ============================================================

embeddings = model.encode(sentences)

print("Embedding shape:", embeddings.shape)

# Expected:
# (4, 384)


# ============================================================
# 5. VIEW ONE EMBEDDING
# ============================================================

print("\nFirst sentence:")
print(sentences[0])

print("\nEmbedding:")
print(embeddings[0])

print("\nEmbedding dimensions:")
print(len(embeddings[0]))


# ============================================================
# 6. COSINE SIMILARITY
# ============================================================

cosine_score = cosine_similarity(
    [embeddings[0]],
    [embeddings[1]]
)[0][0]

print("\nCosine Similarity")
print("-----------------")
print("Sentence 1:", sentences[0])
print("Sentence 2:", sentences[1])
print("Score:", cosine_score)


# ============================================================
# 7. COSINE SIMILARITY MATRIX
# ============================================================

similarity_matrix = cosine_similarity(embeddings)

print("\nCosine Similarity Matrix:")
print(similarity_matrix)


# ============================================================
# 8. DOT PRODUCT
# ============================================================

dot_product = np.dot(
    embeddings[0],
    embeddings[1]
)

print("\nDot Product")
print("-----------")
print("Score:", dot_product)


# ============================================================
# 9. L2 / EUCLIDEAN DISTANCE
# ============================================================

l2_distance = np.linalg.norm(
    embeddings[0] - embeddings[1]
)

print("\nL2 Distance")
print("-----------")
print("Distance:", l2_distance)


# ============================================================
# 10. COMPARE ALL THREE METRICS
# ============================================================

vec1 = embeddings[0]
vec2 = embeddings[1]

cosine = cosine_similarity([vec1], [vec2])[0][0]
dot = np.dot(vec1, vec2)
l2 = np.linalg.norm(vec1 - vec2)

print("\n===================================")
print("SIMILARITY METRIC COMPARISON")
print("===================================")

print("Cosine Similarity :", cosine)
print("Dot Product       :", dot)
print("L2 Distance       :", l2)

print("\nHigher Cosine = More Similar")
print("Higher Dot Product = More Similar")
print("Lower L2 Distance = More Similar")


# ============================================================
# 11. COMPARE ALL SENTENCE PAIRS
# ============================================================

print("\n===================================")
print("SENTENCE PAIR COMPARISON")
print("===================================")

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):

        cosine_score = cosine_similarity(
            [embeddings[i]],
            [embeddings[j]]
        )[0][0]

        l2_distance = np.linalg.norm(
            embeddings[i] - embeddings[j]
        )

        dot_product = np.dot(
            embeddings[i],
            embeddings[j]
        )

        print(f"\nSentence {i + 1}: {sentences[i]}")
        print(f"Sentence {j + 1}: {sentences[j]}")
        print("Cosine Similarity:", cosine_score)
        print("Dot Product:", dot_product)
        print("L2 Distance:", l2_distance)


# ============================================================
# 12. OPENAI EMBEDDING API
# ============================================================
# Requires OPENAI_API_KEY environment variable.
#
# Uncomment this section when your API key is configured.

# from openai import OpenAI
#
# client = OpenAI(
#     api_key=os.getenv("OPENAI_API_KEY")
# )
#
# response = client.embeddings.create(
#     model="text-embedding-3-small",
#     input="What is machine learning?"
# )
#
# openai_embedding = response.data[0].embedding
#
# print("\nOpenAI Embedding")
# print("----------------")
# print("Dimensions:", len(openai_embedding))
# print("First 10 values:", openai_embedding[:10])


# ============================================================
# 13. VOYAGE AI EMBEDDING API
# ============================================================
# Requires VOYAGE_API_KEY environment variable.
#
# Uncomment this section when your API key is configured.

# import voyageai
#
# voyage_client = voyageai.Client(
#     api_key=os.getenv("VOYAGE_API_KEY")
# )
#
# result = voyage_client.embed(
#     ["What is machine learning?"],
#     model="voyage-3"
# )
#
# voyage_embedding = result.embeddings[0]
#
# print("\nVoyage AI Embedding")
# print("-------------------")
# print("Dimensions:", len(voyage_embedding))
# print("First 10 values:", voyage_embedding[:10])


# ============================================================
# 14. SUMMARY
# ============================================================

print("\n===================================")
print("SUMMARY")
print("===================================")

print("""
Sentence Embedding:
Converts text into numerical vectors that represent semantic information.

Cosine Similarity:
Measures the angle/direction between vectors.
Higher value generally means more similar.

Dot Product:
Measures the product of corresponding vector components.
Higher value generally means more similar, but magnitude matters.

L2 Distance:
Measures Euclidean distance between vectors.
Lower value means vectors are closer.

RAG FLOW:

Document
   ↓
Chunks
   ↓
Embeddings
   ↓
Vector Database
   ↓
User Query
   ↓
Query Embedding
   ↓
Similarity Search
   ↓
Relevant Chunks
   ↓
LLM
   ↓
Answer
""")
```
