List

Ordered, mutable collection:

```python
skills = ["Python", "SQL", "ML"]

skills.append("RAG")
print(skills[0])
print(skills)
```

Dictionary

Stores key-value pairs:

```python
student = {
    "name": "Dinesh",
    "age": 20,
    "skill": "Python"
}

print(student["name"])
print(student)
```

Loop

```python
for skill in skills:
    print(skill)

for i in range(5):
    print(i)
```

while Loop

```python
count = 0

while count < 5:
    print(count)
    count += 1
```
