1. Pandas DataFrame

A DataFrame is a table-like data structure in Python.

```python
import pandas as pd

data = {
    "name": ["Dinesh", "Avinash", "Bala"],
    "department": ["AI", "HR", "AI"],
    "salary": [30000, 40000, 35000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head()
df.tail()
df.shape
df.columns
df.info()
df.describe()
```

2. groupby()

Used to group data and perform calculations.

Average salary by department

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["salary"].agg(
    ["count", "mean", "sum", "max", "min"]
)
```

3. merge()

merge() combines two DataFrames, similar to a SQL JOIN.

```python
employees = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Dinesh", "Arun", "Ravi"],
    "department_id": [10, 20, 10]
})

departments = pd.DataFrame({
    "department_id": [10, 20],
    "department_name": ["AI", "HR"]
})

result = pd.merge(
    employees,
    departments,
    on="department_id",
    how="inner"
)

print(result)
```

4. psycopg2

```python
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="intern_db",
    user="postgres",
    password="your_password",
    port=5432
)

cursor = conn.cursor()

cursor.execute("SELECT * FROM employees")

rows = cursor.fetchall()

for row in rows:
    print(row)

cursor.close()
conn.close()
```
